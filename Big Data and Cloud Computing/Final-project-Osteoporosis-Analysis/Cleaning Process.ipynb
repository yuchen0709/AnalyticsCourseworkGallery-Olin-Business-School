```python
import pandas as pd
import numpy as np
```

```python
osteo = pd.read_csv('osteoporosis.csv',sep='$')
```

```python
osteo.head()
```

```python
osteo.replace("Not available", np.nan, inplace=True)
```

```python
osteo2 = osteo.drop(columns=["age","rechal","dechal","age_cod","caseid","exp_dt","country_code","death_dt","indi_hlgt_name","indi_hlt_name","indi_llt_name","indi_meddra_hierarchy_from","indi_meddra_pt","indi_pt","indi_pt_name","indi_soc_name","country_code","reac_hlgt_name","reac_hlt_name","reac_llt_name","reac_meddra_pt","reac_meddra_hierarchy_from","reac_soc_name","gender"])
```

```python
osteo2.dtypes
```

```python
for column in osteo2.select_dtypes(include=['object']).columns:
    osteo2[column] = osteo2[column].str.lower()
```

```python
def drug_name(row):
    if pd.notna(row["prod_ai"]):
        if row["prod_ai"] == "vitamin d nos":
            return "cholecalciferol"
        elif row["prod_ai"] == "alendronate sodium":
            return "alendronic acid"
        elif row["prod_ai"] == "risedronate sodium":
            return "risedronic acid"
        elif row["prod_ai"] == "ibandronate sodium":
            return "ibandronic acid"
        elif row["prod_ai"] == "alendronate sodium\cholecalciferol":
            return "alendronic acid\cholecalciferol"
        else:
            return row["prod_ai"]
    elif pd.isna(row["prod_ai"]):
        if row["drugname"] == "fosamax":
            return "alendronic acid"
        elif row["drugname"] == "forteo":
            return "teriparatide"
        elif row["drugname"] == "prolia":
            return "denosumab"
        elif row["drugname"] == "fosamax plus d":
            return "alendronic acid\cholecalciferol"
        elif row["drugname"] == "alendronate sodium":
            return "alendronic acid"
        elif row["drugname"] == "boniva":
            return "ibandronic acid"
        elif row["drugname"] == "actonel":
            return "risedronic acid"
        elif row["drugname"] == "aclasta":
            return "zoledronic acid"
        elif row["drugname"] == "reclast":
            return "zoledronic acid"
        elif row["drugname"] == "evista":
            return "raloxifene hydrochloride"
        elif row["drugname"] == "calcium (unspecified)":
            return "calcium"
        elif row["drugname"] == "zometa":
            return "zoledronic acid"
        elif row["drugname"] == "calcium":
            return "calcium"
        elif row["drugname"] == "alendronate":
            return "alendronic acid"
        elif row["drugname"] == "alendronate sodium":
            return "alendronic acid"
        elif row["drugname"] == "aredia":
            return "pamidronate disodium"
        elif row["drugname"] == "vitamin d (unspecified)":
            return "cholecalciferol"
        elif row["drugname"] == "vitamin d":
            return "cholecalciferol"
        elif row["drugname"] == "zoledronic acid":
            return "zoledronic acid"
        elif row["drugname"] == "zoledronoc acid":
            return "zoledronic acid"
        elif row["drugname"] == "teriparatide":
            return "teriparatide"
        elif row["drugname"] == "risedronate sodium":
            return "risedronic acid"
        elif row["drugname"] == "alendronic acid":
            return "alendronic acid"
        elif row["drugname"] == "miacalcin":
            return "calcitonin salmon"
        elif row["drugname"] == "cholecalciferol":
            return "cholecalciferol"
        elif row["drugname"] == "vitamin d3":
            return "cholecalciferol"
        elif row["drugname"] == "calcium (unspecified) (+) vitamin d (unspecified)":
            return "calcium carbonate\vitamin d"
        elif row["drugname"] == "forsteo":
            return "teriparatide"
        elif row["drugname"] == "bonalon":
            return "alendronic acid"
        elif row["drugname"] == "alfarol" or row["drugname"] == "alfacalcidol" or row["drugname"] == "onealfa":
            return "alfacalcidol"
        elif row["drugname"] == "calcium carbonate":
            return "calcium carbonate"
        elif row["drugname"] == "bonviva":
            return "ibandronic acid"
        else:
            return row["drugname"]
            
        
        
    
```

```python
osteo2["drug_active_ingredient"] = osteo2.apply(drug_name, axis = 1)
```

```python
def age_groups(row):
    if row["age_in_years"]>= 60:
        return "elder"
    elif row["age_in_years"]>= 18 and row["age_in_years"]<60:
        return "adult"
    elif row["age_in_years"]>= 12 and row["age_in_years"]<18:
        return "teenager"
    elif row["age_in_years"]>=2 and row["age_in_years"]<12:
        return "child"
    elif row["age_in_years"]>=0 and row["age_in_years"]<2:
        return "infant"
    else:
        return np.nan
```

```python
osteo2["age_group"] = osteo2.apply(age_groups, axis = 1)
```

```python
def dosage(row):
    amount = str(row["dose_vbm"])
    if "unk" in amount:
        return np.nan
    else:
        return row["dose_vbm"]
```

```python
osteo2["dosage"] = osteo2.apply(dosage, axis = 1)
```

```python
def weight_kilos(row):
    if row["wt_cod"] == "kg" or row["wt_cod"] == "years":
        return row["wt"]
    elif row["wt_cod"] == "lbs":
        return round(row["wt"]*2.2,2)
    else:
        return np.nan
```

```python
osteo2["weight"] = osteo2.apply(weight_kilos, axis = 1)
```

```python
osteo2["weight_in_kg"] = osteo2["weight"]
```

```python
def dose_form(row):
    dose = str(row["dose_form"])
    if "intravenous" in dose:
        return "IV"
    elif "oral" in dose:
        return "oral"
    elif "capsule" in dose:
        return "oral"
    elif "tablet" in dose:
        return "oral"
    elif "granule" in dose:
        return "oral"
    elif "infusion" in dose:
        return "IV"
    elif "transdermal" in dose:
        return "transdermal"
    elif "intramusc" in dose:
        return "IM"
    elif "patch" in dose:
        return "transdermal"
    elif "ointment" in dose:
        return "topical"
    elif "nasal" in dose:
        return "nasal"
    elif "spray" in dose:
        return "nasal"
    elif "syringe" in dose:
        return "SC"
    elif "inject" in dose:
        return "SC"
    elif "inhal" in dose:
        return "inhaler"
    elif "pen" in dose:
        return "SC"
    elif "gel" in dose:
        return "topical"
    elif "jelly" in dose:
        return "oral"
    elif "pill" in dose:
        return "oral"
    else:
        return np.nan
```

```python
osteo2["drug_delivery"] = osteo2.apply(dose_form, axis = 1)
```

```python
def country_origin(row):
    if row["reporter_country"] == "us":
        return "united states"
    elif row["reporter_country"] == "ca":
        return "canada"
    elif row["reporter_country"] == "jp":
        return "japan"
    elif row["reporter_country"] == "br":
        return "brazil"
    elif row["reporter_country"] == "country not specified":
        return np.nan
    elif row["reporter_country"] == "de":
        return "germany"
    elif row["reporter_country"] == "gb":
        return "united kingdom"
    elif row["reporter_country"] == "fr":
        return "france"
    elif row["reporter_country"] == "es":
        return "spain"
    elif row["reporter_country"] == "au":
        return "australia"
    elif row["reporter_country"] == "dk":
        return "denmark"
    elif row["reporter_country"] == "cn":
        return "china"
    elif row["reporter_country"] == "it":
        return "italy"
    elif row["reporter_country"] == "nl":
        return "netherlands"
    elif row["reporter_country"] == "ie":
        return "ireland"
    elif row["reporter_country"] == "ch":
        return "switzerland"
    elif row["reporter_country"] == "at":
        return "austria"
    elif row["reporter_country"] == "gr":
        return "greece"
    elif row["reporter_country"] == "co":
        return "colombia"
    elif row["reporter_country"] == "il":
        return "israel"
    elif row["reporter_country"] == "mx":
        return "mexico"
    elif row["reporter_country"] == "se":
        return "sweden"
    elif row["reporter_country"] == "be":
        return "belgium"
    elif row["reporter_country"] == "fi":
        return "finland"
    elif row["reporter_country"] == "tw":
        return "taiwan"
    elif row["reporter_country"] == "za":
        return "south africa"
    elif row["reporter_country"] == "kr" or row["reporter_country"] == "korea, republic of":
        return "south korea"
    elif row["reporter_country"] == "pr":
        return "puerto rico"
    elif row["reporter_country"] == "pt":
        return "portugal"
    elif row["reporter_country"] == "in":
        return "india"
    elif row["reporter_country"] == "no":
        return "norway"
    elif row["reporter_country"] == "cz":
        return "czech republic"
    elif row["reporter_country"] == "ar":
        return "argentina"
    elif row["reporter_country"] == "ru":
        return "russia"
    elif row["reporter_country"] == "tr":
        return "turkey"
    else:
        return row["reporter_country"]
```

```python
osteo2["nationality"] = osteo2.apply(country_origin, axis = 1)
```

```python
def outcome_code(row):
    if row["outc_cod"] == "ot":
        return "other"
    elif row["outc_cod"] == "ho":
        return "hospitalization"
    elif row["outc_cod"] == "ds":
        return "disability"
    elif row["outc_cod"] == "de":
        return "death"
    elif row["outc_cod"] == "lt":
        return "life-threatening"
    elif row["outc_cod"] == "ri":
        return "required intervention"
    elif row["outc_cod"] == "ca":
        return "congenital anomaly"
    else:
        return np.nan
```

```python
osteo2["outcome"] = osteo2.apply(outcome_code, axis = 1)
```

```python
osteo2["adverse_reaction"] = osteo2["pt"]
```
